```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from GaPFlow.problem import Problem
from utils.plotting import plot_overview_1d, plot_overview_2d, plot_midsection_2d
from utils.topography import (
    regen_conv_slider_pocket_1d,
    regen_conv_slider_pocket_2d,
    regen_twin_parabolic_slider,
    regen_twin_parabolic_slider_id,
)

DATA_DIR = Path('../../tests/data')
```

---
## Convergent slider with pocket — 1D

Linear converging wedge (1.1→1.0 µm) with a 0.4 µm rectangular pocket at x∈[4,10] mm.  
Periodic in y (effectively 1D). Dirichlet p=1e5 Pa at inlet and outlet.

**Reference:** Bertocchi et al., Trib. Int. 67 (2013), 61–71.

<img src="./figures/Geometry_ConvSliderPocket.png"
     alt="Conv. slider with pocket geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/conv_slider_pocket_1D.yaml)**

```python
problem_1d = Problem.from_yaml('../../tests/configs/conv_slider_pocket_1D.yaml')
regen_conv_slider_pocket_1d(problem_1d, Nx_geo=50)
problem_1d.run()
```

```python
plot_overview_1d(
    problem_1d,
    title='Conv. slider + pocket 1D (Bertocchi 2013)',
    ref_csv=DATA_DIR / 'ref_conv_slider_pocket_1D.csv',
    ref_label='Bertocchi',
    ref_x_scale=0.020,
    ref_p_scale=1e6,
    p_unit='MPa', p_divisor=1e6,
)
```

## Twin parabolic slider non-identical — 1D

Two asymmetric parabolic sliders (hmin₁=30 µm, hmin₂=35 µm) with a flat separator and flat outlet region.
Dirichlet ρ at ambient at both ends, periodic in y.

**Reference:** Bayada et al. (Fig. 7).

<img src="./figures/Geometry_TwinParabolicSlider.png"
     alt="Twin parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/twin_parabolic_slider.yaml)**

```python
problem_twin = Problem.from_yaml('../../tests/configs/twin_parabolic_slider.yaml')
regen_twin_parabolic_slider(problem_twin)
problem_twin.run()
```

```python
plot_overview_1d(
    problem_twin,
    title='Twin parabolic slider uneven',
    ref_csv=DATA_DIR / 'ref_twin_parabolic_slider.csv',
    ref_label='Bayada',
    ref_x_scale=0.09,
    ref_p_scale=1e5,
    p_unit='bar', p_divisor=1e5,
)
```

High inlet pressure p_in=3.36e5 Pa, p_out≈0 Pa, periodic in y.

**Reference:** Sahlin et al. / Giacopini et al., J. Tribology 132 (2010), 041702.

<img src="./figures/Geometry_TwinParabolicSliderIdentical.png"
     alt="Twin identical parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/twin_parabolic_slider_id.yaml)**

```python
problem_twin_id = Problem.from_yaml('../../tests/configs/twin_parabolic_slider_id.yaml')
regen_twin_parabolic_slider_id(problem_twin_id)
problem_twin_id.run()
```

```python
plot_overview_1d(
    problem_twin_id,
    title='Twin parabolic slider identical',
    ref_csv=DATA_DIR / 'ref_twin_parabolic_slider_id.csv',
    ref_label='Sahlin',
    ref_x_scale=problem_twin_id.grid['Lx'],
    ref_p_scale=1e6,
    p_unit='MPa', p_divisor=1e6,
)
```
